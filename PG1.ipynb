```python
import pandas as pd


data = pd.DataFrame([
    ['Sunny','Warm','Normal','Strong','Warm','Same','Yes'],
    ['Sunny','Warm','High','Strong','Warm','Same','Yes'],
    ['Rainy','Cold','High','Strong','Warm','Change','No'],
    ['Sunny','Warm','High','Strong','Cool','Change','Yes']
], columns=['Sky','AirTemp','Humidity','Wind','Water','Forecast','EnjoySport'])

print(data)


def find_s(data):
    h = ['0'] * 6

    for _, row in data.iterrows():
        if row['EnjoySport'] == 'Yes':
            for j in range(6):
                if h[j] == '0':
                    h[j] = row.iloc[j]
                elif h[j] != row.iloc[j]:
                    h[j] = '?'
    return h

print("\nFind-S Hypothesis:", find_s(data))



def candidate_elimination(data):
    S = ['0'] * 6
    G = ['?'] * 6

    for _, row in data.iterrows():
        if row['EnjoySport'] == 'Yes':
            for j in range(6):
                if S[j] == '0':
                    S[j] = row.iloc[j]
                elif S[j] != row.iloc[j]:
                    S[j] = '?'
        else:
            for j in range(6):
                if S[j] == row.iloc[j]:
                    S[j] = '?'

    return S, G

S, G = candidate_elimination(data)

print("\nCandidate Elimination:")
print("S =", S)
print("G =", G)
```
